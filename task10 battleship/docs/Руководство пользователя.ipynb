# Руководство для пользователя

## Приложение "Морской бой"

### 1. Введение
Приложение позволяет играть в классическую игру «Морской бой» против компьютера с тремя уровнями сложности, вести таблицу рекордов и настраивать параметры игры.

---

### 2. Установка и запуск
1. Распакуйте папку с приложением.
2. Убедитесь, что на компьютере установлен Python 3.8 или выше.
3. Установите зависимости: ```pip install PySide6```
4. Откройте терминал в папке с приложением.
5. Запустите программу: ```python main.py```

---

### 3. Главное меню
При запуске открывается главное меню с кнопками:

| Кнопка | Назначение |
|--------|------------|
| **Новая игра** | Начать игру против компьютера |
| **Настройки** | Открыть окно настройки сложности |
| **Рекорды** | Показать таблицу лучших результатов |
| **Выход** | Закрыть приложение |

### 4. Игровой процесс

#### 4.1. Расстановка кораблей
- Корабли расставляются автоматически при запуске игры.
- Кнопка «Переставить корабли» генерирует новую случайную расстановку.
- Кнопка доступна только до первого выстрела.

<center>

</center>

#### 4.2. Игровое поле
- Слева — ваше поле с кораблями.
- Справа — поле противника (корабли скрыты).

#### 4.3. Выстрелы
Кликните по клетке на поле противника для выстрела.

Результаты обозначаются цветами:

| Цвет | Значение |
|------|----------|
| Белый | Промах |
| Красный | Попадание |
| Тёмно-красный | Корабль потоплен |

#### 4.4. Правило хода
- Промах — ход переходит к компьютеру.
- Попадание — игрок ходит снова.
- Потопление — игрок ходит снова.

#### 4.5. Завершение игры
Игра заканчивается, когда:
- Игрок потопил все корабли противника (победа).
- Компьютер потопил все корабли игрока (поражение).

</center>

При победе появится окно для ввода имени — результат сохранится в таблицу рекордов.

### 5. Настройки

<center>

</center>

В настройках можно выбрать уровень сложности:

| Уровень | Стратегия компьютера |
|---------|---------------------|
| **Матрос** | Случайные выстрелы |
| **Боцман** | Смешанная стратегия |
| **Капитан** | Оптимальная стратегия |

Выбранная сложность сохраняется автоматически и загружается при следующем запуске.

---

### 7. Часто задаваемые вопросы

**В: Как изменить сложность?**
О: Откройте «Настройки», выберите уровень, нажмите «ОК».

**В: Как сбросить рекорды?**
О: Удалите файл ```records.json``` из папки приложения.

**В: Почему корабли расставлены случайно?**
О: Для упрощения игры. Нажмите «Переставить» для новой расстановки.

**В: Можно ли расставить корабли вручную?**
О: В текущей версии — нет, только случайная расстановка.
