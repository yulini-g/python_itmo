## Приложение "Фермерские рынки США"

### 1. Введение
Приложение позволяет просматривать, искать, сортировать и оценивать фермерские рынки США.

### 2. Установка и запуск
1. Распакуйте папку с приложением.
2. Убедитесь, что на компьютере установлен Python 3.8 или выше.
3. Откройте папку с приложением.
4. **Способ 1 (рекомендуемый):** дважды кликните на `setup.bat`.
5. **Способ 2:** откройте терминал (`cmd`) и выполните:

### 3. Главное меню
После запуска вы увидите:

1. Просмотр всех рынков
2. Поиск по городу и штату
3. Поиск по ZIP
4. Поиск по удаленности
5. Сортировка рынков
6. Удалить рынок
7. Выход
   
==================================================

Выберите действие (1-7).
Введите номер пункта и нажмите **Enter**.

### 4. Функции приложения

#### 4.1. Просмотр всех рынков (пункт 1)
- Показывает все рынки по 10 на странице.
- **Навигация:**
  - `1` — следующая страница
  - `2` — предыдущая страница
  - `3` — выбрать рынок для просмотра деталей
  - `0` — выход в главное меню

#### 4.2. Поиск по городу и штату (пункт 2)
1. Введите город или `0` (любой).
2. Введите штат или `0` (любой).
3. Программа покажет список найденных рынков.
4. Вы можете выбрать рынок для просмотра деталей.

#### 4.3. Поиск по ZIP-коду (пункт 3)
1. Введите реально существующий 5-значный ZIP-код.
2. Программа покажет все рынки с этим индексом.

#### 4.4. Поиск по удаленности (пункт 4)
1. Введите ZIP-код, вокруг которого будет происходить поиск.
2. Введите радиус поиска в милях.
3. Программа покажет рынки в пределах указанного радиуса.

#### 4.5. Сортировка рынков (пункт 5)
1. Выберите параметр сортировки:
   - `1` — Город и штат
   - `2` — Название рынка
   - `3` — Город
   - `4` — Штат
   - `5` — ZIP-код
   - `6` — Рейтинг
2. Выберите порядок:
   - `1` — по возрастанию
   - `2` — по убыванию
3. Программа покажет отсортированный список.

#### 4.6. Удалить рынок (пункт 6)
1. Выберите рынок из списка.
2. Подтвердите удаление.
3. Рынок и все его отзывы будут удалены навсегда.

#### 4.7. Выход (пункт 7)
Завершает работу программы.

### 5. Детальный просмотр рынка
При выборе рынка (через пункт `3` в любом списке) вы увидите:
- Название
- FMID (уникальный идентификатор)
- Адрес
- Город, штат, ZIP
- Сезон работы
- Количество отзывов
- Средний рейтинг
- Возможность посмотреть отзывы
- Возможность оставить свой отзыв

### 6. Ошибки и их устранение

| Сообщение | Что делать |
|-----------|------------|
| `Файл не найден` | Убедитесь, что `database.csv` находится в папке с программой |
| `Неверная команда` | Введите число из предложенного списка |
| `Нет данных для отображения` | По вашему запросу ничего не найдено |
| `Ошибка: рейтинг должен быть от 1 до 5` | Введите число от 1 до 5 |

### 8. Возможные вопросы

**Вопрос:** Что делать, если приложение не запускается?  
**Ответ:** Убедитесь, что установлен Python. Скачайте с python.org.

**Вопрос:** Можно ли указать другой файл с данными?  
**Ответ:** Нет, программа всегда использует файл `database.csv`.

**Вопрос:** Где хранятся отзывы?  
**Ответ:** В файле `reviews.json` в той же папке.
