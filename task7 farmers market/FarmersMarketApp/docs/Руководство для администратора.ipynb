## Приложение "Фермерские рынки США"

### 1. Системные требования
- **Операционная система:** Windows 7/10/11
- **Python:** версия 3.8 или выше (скачать: python.org)
- **Дополнительные модули:** не требуются (используются только встроенные модули: `csv`, `json`, `math`)

---

### 2. Автоматическое развертывание

#### 2.1. На клиентской машине
1. Скопируйте папку с приложением на компьютер пользователя.
2. Пользователь запускает `setup.bat` двойным кликом.
3. Скрипт проверяет наличие Python и запускает программу.

#### 2.2. Проверка успешного развертывания
- Приложение открывается с главным меню.
- В папке создаётся файл `reviews.json` (если его не было).

---

### 3. Ручное развертывание (если автоматическое не сработало)

1. Установите Python 3.8+ с официального сайта.
2. Откройте командную строку (`cmd`) в папке с приложением.
3. Запустите:
   ```cmd
   python "Farmers Market.py"

### 5. Формат входных данных (database.csv)

- Файл должен быть в формате CSV с заголовками.
- Минимально необходимые колонки: FMID, MarketName, city, State, zip, x, y.
- Другие колонки загружаются автоматически.
- Если колонки отсутствуют, программа может работать некорректно.

Пример допустимого файла:
    
```csv```
    
```FMID,MarketName,city,State,zip,x,y```
    
```1018965,Best Market,Springfield,CA,12345,-122.5,37.8```


## 6. Эксплуатация и поддержка
### 6.1. Резервное копирование
- Регулярно создавайте копии файлов database.csv и reviews.json.
- При потере данных — замените их на копии.

### 6.2. Обновление данных
- Замените database.csv на новый файл.
- При следующем запуске программа загрузит новые данные.

### 6.3. Очистка отзывов
Удалите содержимое reviews.json или удалите сам файл.

### 7. Ошибки и их устранение

| Проблема | Решение |
|-----------|------------|
| `Python not found` | Установить Python 3.8+ |
| `FileNotFoundError: database.csv` | Поместить database.csv в папку с программой |
| `PermissionError` | Запустить от имени администратора |
| `KeyError: 'FMID'` | Проверить заголовки в database.csv |

## 8. Конфигурация

Программа использует жёстко заданные имена файлов:

- `database.csv` — данные о рынках.
- `reviews.json` — отзывы.

Для изменения путей требуется правка кода.

## 9. Обновление приложения

1. Замените файл Farmers Market.py на новую версию.
2. Сохраните файлы database.csv и reviews.json (если в них есть данные).
3. Запустите программу.
