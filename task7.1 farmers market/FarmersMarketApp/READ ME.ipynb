# Фермерские рынки США

Приложение для просмотра, поиска, сортировки и оценки фермерских рынков США.

---

## Быстрый старт

1. Установите Python 3.8+
2. Запустите `setup.bat` (Windows) или `setup.sh` (Linux/macOS)
3. Следуйте инструкциям в меню

---

## Файлы

- `Farmers Market.py` — основной код
- `database.csv` — данные о рынках
- `reviews.json` — отзывы пользователей

---

## Требования

- Python 3.8+
- Модули: csv, json, math (все встроенные)
- ОС: Windows
