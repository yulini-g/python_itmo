# Руководство пользователя

## Приложение "Морской бой" (клиент-серверная версия)

### 1. Введение
Приложение позволяет играть в классическую игру «Морской бой» против удалённого сервера, который выступает в роли второго игрока. Есть три уровня сложности, таблица рекордов и настройка параметров игры.

---

### 2. Установка и запуск

#### 2.1. Запуск сервера (на удалённом компьютере)
1. Убедитесь, что на компьютере установлен Python 3.8 или выше.
2. Установите зависимости: ```pip install PySide6```
3. Откройте терминал в папке с приложением.
4. Запустите сервер: ```python server.py```
5. Сервер готов принимать подключения.

#### 2.2. Запуск клиента (на компьютере игрока)
1. Убедитесь, что на компьютере установлен Python 3.8 или выше.
2. Установите зависимости: ```pip install PySide6```
3. Откройте терминал в папке с приложением.
4. Запустите программу: ```python main.py```
5. Приложение откроется на весь экран.

---

При запуске открывается главное меню с кнопками:

| Кнопка | Назначение |

### 4. Игровой процесс

#### 4.1. Подключение к серверу
При нажатии «Новая игра» клиент подключается к серверу. Сервер создаёт отдельный сеанс для каждого игрока.

#### 4.2. Расстановка кораблей
- Корабли расставляются автоматически при запуске игры.
- Расстановка отправляется на сервер.
- Кнопка «Переставить корабли» генерирует новую случайную расстановку.
- Кнопка доступна только до первого выстрела.

<center>

</center>

#### 4.3. Игровое поле
- Слева — ваше поле с кораблями.
- Справа — поле противника (корабли скрыты).
- Под полем противника отображается счётчик ходов.

#### 4.4. Выстрелы
Кликните по клетке на поле противника для выстрела.

Результаты обозначаются цветами:

| Цвет | Значение |
|------|----------|
| Белый | Промах |
| Красный | Попадание |
| Тёмно-красный | Корабль потоплен |

#### 4.5. Правило хода
- Промах — ход переходит к серверу.
- Попадание — игрок ходит снова.
- Потопление — игрок ходит снова.

Во время хода сервера поле противника блокируется.

#### 4.6. Завершение игры
Игра заканчивается, когда:
- Игрок потопил все корабли противника (победа).
- Сервер потопил все корабли игрока (поражение).

### 7. Часто задаваемые вопросы

**В: Как изменить сложность?**
О: Откройте «Настройки», выберите уровень, нажмите «ОК».

**В: Как сбросить рекорды?**
О: Удалите файл ```records.json``` из папки приложения.

**В: Почему корабли расставлены случайно?**
О: Для упрощения игры. Нажмите «Переставить» для новой расстановки.

**В: Можно ли расставить корабли вручную?**
О: В текущей версии — нет, только случайная расстановка.

**В: Как подключиться к другому серверу?**
О: Измените адрес в файле ```gui/network_client.py``` (по умолчанию localhost:5555).

**В: Почему не подключается?**
О: Убедитесь, что сервер запущен и доступен по сети.
