# Руководство для администратора

## Приложение "Морской бой" (клиент-серверная версия)

### 1. Системные требования

| Компонент | Требование |
|-----------|------------|
| Операционная система | Windows 10/11, Linux (Ubuntu 20.04+) |
| Python | 3.8 или выше |
| Дополнительные библиотеки | PySide6 |
| Сеть | Локальная сеть или интернет |
| Свободное место на диске | 100 МБ |
| Разрешение экрана | 1024×768 или выше |

---

### 2. Автоматическое развертывание

#### 2.1. Запуск сервера (Windows)

1. Скопируйте папку с приложением на серверный компьютер.
2. Откройте терминал (cmd) в папке с приложением.
3. Выполните:
```
pip install PySide6
python server.py
```
Сервер запущен и ожидает подключений.

#### 2.2. Запуск сервера (Linux)

1. Скопируйте папку с приложением на серверный компьютер.
2. Откройте терминал в папке с приложением.
3. Выполните:
```
pip3 install PySide6
python3 server.py
```

#### 2.3. Запуск клиента

На компьютере игрока:
```
pip install PySide6
python main.py
```

#### 2.4. Проверка успешного развертывания

- Сервер выводит: «Сервер морского боя запущен! Ожидание игроков...»
- Клиент открывается с главным меню.
- При нажатии «Новая игра» клиент подключается к серверу.
- В папке создаётся файл ```config.json``` (если его не было).
- В папке создаётся файл ```records.json``` (если его не было).

---

### 3. Ручное развертывание

Если автоматическое развертывание не сработало:

1. Установите Python 3.8+ с официального сайта python.org.
2. Откройте командную строку в папке с приложением.
3. Установите зависимости: ```pip install -r requirements.txt```
4. Запустите сервер: ```python server.py```
5. На клиентской машине запустите: ```python main.py```

### 5. Файлы конфигурации

#### 5.1. config.json

Содержит настройки приложения.

Формат:
```
{
"difficulty": "medium"
}
```
Возможные значения difficulty:
- easy (Матрос)
- medium (Боцман)
- hard (Капитан)

#### 5.2. records.json

Содержит таблицу рекордов по уровням сложности.

Формат:
```
{
"easy": [
{"name": "Игрок", "score": 45}
],
"medium": [],
"hard": []
}
```
- score — количество ходов (меньше = лучше).
- Максимум 10 записей на каждый уровень.

#### 5.3. Настройка сети

Адрес и порт сервера задаются в файле ```gui/network_client.py```:

```def __init__(self, host='localhost', port=5555):```

И в файле server.py:

```server.bind(('localhost', 5555))```

Для подключения с удалённого компьютера замените localhost на IP-адрес сервера.

---

### 6. Многопоточность

Сервер использует модуль threading для обслуживания нескольких клиентов одновременно.

Каждый клиент получает отдельный поток:

```
thread = threading.Thread(target=handle_client, args=(client, address))
thread.start()
```

Ограничение на количество одновременных подключений задаётся в ```server.listen(5)``` — максимум 5 клиентов в очереди.

### 7. Эксплуатация и поддержка

#### 7.1. Резервное копирование

Регулярно создавайте копии файлов:
- config.json
- records.json

При потере данных — замените файлы на резервные копии.

#### 7.2. Очистка рекордов

Удалите содержимое records.json или удалите сам файл.

При следующем запуске файл создастся автоматически с пустой таблицей.

#### 7.3. Сброс настроек

Удалите файл config.json.

При следующем запуске будут применены значения по умолчанию (сложность: Боцман).

#### 7.4. Мониторинг сервера

Сервер выводит в консоль:
- Подключения клиентов
- Выстрелы клиентов
- Ходы сервера
- Отключения клиентов

### 8. Ошибки и их устранение

| Проблема | Решение |
|----------|---------|
| ModuleNotFoundError: PySide6 | Установить: pip install PySide6 |
| ModuleNotFoundError: gui | Запускать из корневой папки приложения |
| ConnectionRefusedError | Сервер не запущен или неверный порт |
| ConnectionResetError | Клиент закрыл соединение (нормально) |
| Файл config.json повреждён | Удалить файл, он создастся заново |
| Файл records.json повреждён | Удалить файл, он создастся заново |
| PermissionError при записи | Проверить права на запись в папку |
| Окно не помещается на экран | Уменьшить разрешение экрана |
| Сервер не принимает подключения | Проверить брандмауэр, открыть порт 5555 |

---

### 9. Обновление приложения

1. Замените файлы приложения на новую версию.
2. Сохраните config.json и records.json (если в них есть данные).
3. Обновите зависимости: ```pip install --upgrade PySide6```
4. Перезапустите сервер: ```python server.py```
5. Запустите клиент: ```python main.py```

### 12. Безопасность

- Сервер не использует аутентификацию — подключаться может любой клиент в сети.
- Для развёртывания в публичной сети рекомендуется добавить проверку IP-адресов или токены.
- Порт 5555 должен быть открыт в брандмауэре сервера.
- При работе в локальной сети дополнительная настройка не требуется.
