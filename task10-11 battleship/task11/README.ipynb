# Морской бой (клиент-серверная версия)

Игра «Морской бой» с клиент-серверной архитектурой. Игрок подключается к удалённому серверу, который выступает в роли второго игрока. Сервер многопоточный, поддерживает несколько одновременных подключений. Три уровня сложности, таблица рекордов, настраиваемые параметры.

---

## Быстрый старт

### Запуск сервера

1. Установите Python 3.8+
2. Установите зависимости: pip install PySide6
3. Запустите сервер: python server.py

### Запуск клиента

1. Установите Python 3.8+
2. Установите зависимости: pip install PySide6
3. Запустите клиент: python main.py

---

## Файлы

- server.py — многопоточный сервер
- main.py — точка входа клиента
- gui/ — графический интерфейс
- game/ — логика игры и ИИ
- config.json — настройки (создаётся автоматически)
- records.json — таблица рекордов (создаётся автоматически)
- requirements.txt — зависимости

---

## Требования

- Python 3.8+
- PySide6
- ОС: Windows 10/11, Linux
- Сеть: локальная или интернет
